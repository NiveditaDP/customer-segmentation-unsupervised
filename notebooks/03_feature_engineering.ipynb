```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import os

# load processed data
df = pd.read_csv("../data/processed/processed_customer_data.csv")

# scaling
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)

# PCA (2 components for visualization)
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# save scaled data and PCA
os.makedirs("../results/pca_outputs", exist_ok=True)
np.save("../results/pca_outputs/scaled_data.npy", X_scaled)
np.save("../results/pca_outputs/pca_data.npy", X_pca)
```

```python
import pandas as pd

# Load processed data
df = pd.read_csv("../data/processed/processed_customer_data.csv")

# Feature Engineering

df['Engagement Score'] = df['Interactions with Customer Service'] * df['Insurance Products Owned']

df['Income per Product'] = df['Income Level'] / (df['Insurance Products Owned'] + 1)

df['Age Group'] = pd.cut(df['Age'], bins=3, labels=[0,1,2])

df['High Value'] = (df['Income Level'] > df['Income Level'].mean()).astype(int)

# Save updated dataset
df.to_csv("../data/processed/featured_customer_data.csv", index=False)

df.head()
```

```python
#FEATURE ENGINEERING:

- Created Engagement Score to measure customer interaction level.
- Created Income per Product to evaluate spending efficiency.
- Categorized Age into groups for segmentation.
- Identified high-value customers based on income.

These features help improve clustering performance and business understanding.
```
