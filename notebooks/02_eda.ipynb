```python
raw_df = pd.read_csv("../data/raw/customer_dataset.csv")
```

```python
#MULTIVARIATE
#CALCULATED FROM RAW DATA
plt.figure(figsize=(6,4))
sns.countplot(x='Gender', data=raw_df)
plt.title("Gender Distribution")
plt.show()

```

#Gender Distribution Insights:
The dataset shows that the number of male customers is higher compared to female customers.
This indicates a possible gender imbalance in the customer base.

```python
##MULTIVARIATE
#USING PROCESSED DATA FOR CALCULATION
plt.figure(figsize=(12,6))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
plt.title("Feature Correlation Heatmap")
plt.show()
```

```python
#MULTIVARIATE
#USED RAW DATA FOR CALCULATION
plt.figure(figsize=(8,5))
sns.countplot(x='Education Level', data=raw_df)
plt.title("Education Level Distribution")
plt.xticks(rotation=45)
plt.show()

```

#Education Level Insight:
The majority of customers have an Associate Degree, indicating that mid-level education is dominant in the dataset.
This suggests that customer behavior and purchasing patterns may be influenced by this education group.


```python
#BIVARIATE
plt.figure(figsize=(6,4))
sns.scatterplot(x=raw_df['Age'], y=raw_df['Income Level'])
plt.title("Age vs Income")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x=raw_df['Gender'], y=raw_df['Income Level'])
plt.title("Gender vs Income")
plt.show()
```

#PATTERN DISCOVERY

- Majority of customers fall into mid-range income categories.
- Male customers dominate the dataset.
- Most customers have an Associate Degree.
- There is no strong correlation between features, indicating independent behavior.
- Customer characteristics vary widely, suggesting the need for clustering techniques.
