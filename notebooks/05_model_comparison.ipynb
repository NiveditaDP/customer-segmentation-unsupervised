```python
import pandas as pd

df = pd.read_csv("../data/processed/featured_customer_data.csv")
```

```python
X = df.select_dtypes(include=['int64', 'float64'])

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
#K MEANS
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=2, random_state=42)
df['KMeans_Cluster'] = kmeans.fit_predict(X_scaled)
```

```python
#HIERARCHICAL
from sklearn.cluster import AgglomerativeClustering

hc = AgglomerativeClustering(n_clusters=2)
df['Hierarchical_Cluster'] = hc.fit_predict(X_scaled)
```

```python
#DBSCAN
from sklearn.cluster import DBSCAN

db = DBSCAN(eps=0.5, min_samples=5)
df['DBSCAN_Cluster'] = db.fit_predict(X_scaled)
```

```python
#GMM
from sklearn.mixture import GaussianMixture

gmm = GaussianMixture(n_components=2, random_state=42)
df['GMM_Cluster'] = gmm.fit_predict(X_scaled)
```

```python
#MODEL COMPARISON
from sklearn.metrics import silhouette_score

models = {
    "KMeans": df['KMeans_Cluster'],
    "Hierarchical": df['Hierarchical_Cluster'],
    "DBSCAN": df['DBSCAN_Cluster'],
    "GMM": df['GMM_Cluster']
}

for name, labels in models.items():
    if len(set(labels)) > 1:
        score = silhouette_score(X_scaled, labels)
        print(f"{name} Silhouette Score: {score}")
    else:
        print(f"{name} could not form valid clusters")
```

```python
Model Comparison:

KMeans and GMM achieved the highest silhouette score (0.118), indicating similar clustering performance.

Hierarchical clustering showed slightly lower performance.

DBSCAN failed to form meaningful clusters, as most points were classified as noise.

Therefore, KMeans is selected as the final model due to its simplicity, efficiency, and interpretability.
```
